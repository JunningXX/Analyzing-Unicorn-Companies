# Project Overview: The Global Unicorn Landscape (2019–2022)

## Introduction

A *unicorn* is a privately held startup company with a valuation of over $1 billion. This dataset provides a comprehensive look into the global unicorn landscape during one of the most volatile and fascinating periods in modern financial history: the steady growth of 2019-2020, the unprecedented venture capital boom of 2021, and the severe market correction of 2022.

This project analyses a relational database containing over 1,000 unicorn companies globally. The data is structured across four key tables:

**dates**: Tracks the timeline from the year the company was founded to the exact date it reached unicorn status.

**funding**: Contains the financial metrics, including total funding raised, current valuation, and a list of key venture capital investors.

**industries**: Categorises the specific market sector (e.g., Fintech, Edtech, SaaS).

**companies**: Details the geographic footprint (country, continent, city) of each startup.

```python
-- Median time taken to unicorn group by industries

WITH temp_table AS (
	SELECT 
	d.company_id,
	EXTRACT(YEAR FROM d.date_joined) AS year_joined,
	d.year_founded,
	c.company,
	c.continent,
	i.industry,
	EXTRACT(YEAR FROM d.date_joined) - d.year_founded AS years_taken
FROM dates AS d
INNER JOIN companies AS c
USING(company_id)
INNER JOIN industries AS i
USING(company_id)
	)
SELECT 
	industry,
	PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY years_taken) AS median_years_to_unicorn
FROM temp_table
GROUP BY 
    industry
ORDER BY 
    median_years_to_unicorn ASC;

-- Median time taken to unicorn group by continent

WITH temp_table AS (
	SELECT 
	d.company_id,
	EXTRACT(YEAR FROM d.date_joined) AS year_joined,
	d.year_founded,
	c.company,
	c.continent,
	i.industry,
	EXTRACT(YEAR FROM d.date_joined) - d.year_founded AS years_taken
FROM dates AS d
INNER JOIN companies AS c
USING(company_id)
INNER JOIN industries AS i
USING(company_id)
	)
SELECT 
	continent,
	PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY years_taken) AS median_years_to_unicorn
FROM temp_table
GROUP BY 
    continent
ORDER BY 
    median_years_to_unicorn ASC;
```

## Question 2: The "Investor Portfolio" Breakdown

>  ### Who are the most prolific investors, and what is the total combined valuation of all the unicorns they have backed?

### 📝 Key Insights for Table 3:

* Volume vs. Value (The Sequoia China Anomaly): While Accel takes the No.1 spot for the sheer number of unicorns backed (60), their total portfolio valuation is only \\$216B. Compare that to Sequoia Capital China, which backed fewer companies (48), but has a staggering total valuation of \\$473B - more than double Accel's! This indicates that Sequoia China is investing in massive "mega-unicorns" (like ByteDance/TikTok), whereas Accel takes a wider "spread" approach.

```python
WITH investor_portfolio AS (
    SELECT 
        company_id,
        valuation,
		-- Chop the string into a list
        -- Explode the list into multiple rows
        UNNEST(STRING_TO_ARRAY(select_investors, ', ')) AS investor
    FROM 
        funding
    -- filter out rows where investors are completely missing
    WHERE 
        select_investors IS NOT NULL
)
SELECT 
    investor,
    COUNT(DISTINCT company_id) AS num_unicorns_backed,
    ROUND(SUM(valuation) / 1000000000, 2) AS total_portfolio_valuation_billions
FROM 
    investor_portfolio
GROUP BY 
    investor
ORDER BY 
    num_unicorns_backed DESC
LIMIT 10;
```

## Question3: Year-Over-Year (YoY) Growth 

> ### Which countries are the fastest-growing emerging hubs for unicorns? Compare the number of new unicorns a country produced from 2020 to 2022.

### 📝 Key Insights: Global Unicorn Trends (2020 - 2022)

* The 2021 Venture Capital Bubble: The year 2021 was a global anomaly for startup valuations. Every single country on the list experienced triple or quadruple-digit percentage growth. Canada was the most extreme emerging outlier, growing an astonishing 1,400% (jumping from 1 unicorn in 2020 to 15 in 2021). The United States remained the undisputed king of volume, minting a staggering 303 new unicorns in a single year.
<br>

* The Severe 2022 Market Correction: The data shows a brutal and universal market cool-down in 2022. With the exception of Ireland (which remained flat), every country experienced negative YoY growth ranging from -50% to -95%. China was hit the hardest by this contraction, dropping -95.45% (plummeting from 44 new unicorns in 2021 down to just 2 in 2022).
<br>
* Identifying the True "Emerging Hubs": To find the real emerging hubs, we must look at which countries survived the 2022 crash with a higher baseline than they had in 2020.
     - India: Proved to be a highly resilient emerging market. Despite the 2022 crash, they still minted 11 unicorns—noticeably higher than their 2020 baseline of 7.
<br>
     - France & Canada: Both of these nations proved they are growing tech hubs. France doubled its pre-bubble baseline (going from 2 in 2020 to 4 in 2022), and Canada tripled its baseline (1 in 2020 to 3 in 2022).

```python
WITH temp_table AS (
    SELECT 
        d.company_id,
        EXTRACT(YEAR FROM d.date_joined) AS year_joined,
        d.year_founded,
        c.company,
        c.country,
        i.industry
    FROM dates AS d
    INNER JOIN companies AS c USING(company_id)
    INNER JOIN industries AS i USING(company_id)
),
yearly_count AS (
    SELECT 
        country,
        year_joined,
        COUNT(company) AS num_unicorns
    FROM temp_table 
    WHERE year_joined IN (2020, 2021, 2022)
    GROUP BY country, year_joined
),
filtered_countries AS (
    SELECT 
        country,
        year_joined,
        num_unicorns,
        COUNT(year_joined) OVER (PARTITION BY country) AS years_present
    FROM yearly_count
),
appeared_3_years AS (
    SELECT 
        country,
        year_joined,
        num_unicorns
    FROM filtered_countries
    WHERE years_present = 3 
),
growth_calculations AS (
    SELECT 
        country,
        year_joined,
        num_unicorns,
        ROUND(
            ((num_unicorns - LAG(num_unicorns, 1) OVER (PARTITION BY country ORDER BY year_joined)) 
            / 
            CAST(LAG(num_unicorns, 1) OVER (PARTITION BY country ORDER BY year_joined) AS NUMERIC)) * 100
        , 2) AS yoy_growth_pct
    FROM appeared_3_years
)
SELECT 
    country,
    -- 2020 Data
    MAX(CASE WHEN year_joined = 2020 THEN num_unicorns END) AS unicorns_2020,
    
    -- 2021 Data
    MAX(CASE WHEN year_joined = 2021 THEN num_unicorns END) AS unicorns_2021,
    MAX(CASE WHEN year_joined = 2021 THEN yoy_growth_pct END) AS growth_2020_to_2021_pct,
    
    -- 2022 Data
    MAX(CASE WHEN year_joined = 2022 THEN num_unicorns END) AS unicorns_2022,
    MAX(CASE WHEN year_joined = 2022 THEN yoy_growth_pct END) AS growth_2021_to_2022_pct
FROM 
    growth_calculations
GROUP BY 
    country
ORDER BY 
    country ASC;
```

## Question 4: The "Capital Efficiency" Ratio 

> ### Which companies did the "most with the least"? Find the companies with the highest Valuation-to-Funding ratio, then categorise them into efficiency tiers.

### 📝 Key Insights: The Top 10 Capital Efficient Unicorns

* The "Zapier Anomaly" (Extreme Bootstrapping): Zapier is in a league of its own. With a ratio of 4,000.00, they generated four thousand dollars in valuation for every single dollar of funding they took. The gap between #1 (Zapier at 4000) and #2 (Dunamu at 126) is staggering. This perfectly illustrates the power of "bootstrapping"—building a highly profitable product using customer revenue rather than relying on venture capital.
<br>

* The "100x Club": The top five companies on this list (Zapier, Dunamu, Workhuman, CFGI, Manner) all achieved a ratio of 100 or greater. In the venture capital world, returning 10x on an investment is considered a massive win. These companies are generating \\$100+ of enterprise value for every \\$1 raised, making them some of the most lucrative investments in the global ecosystem.
<br>
* Hardware Can Compete: A common assumption in tech is that only SaaS (Software as a Service) companies can be highly capital efficient because software is cheap to duplicate. However, DJI Innovations (Drones) and GalaxySpace (Aerospace) completely break that rule. Despite being hardware manufacturers with massive supply chain and R&D costs, they still managed to crack the top 10 with ratios over 70.
<br>
* Global Diversity: Highly efficient unicorns are not restricted to Silicon Valley. Looking at this list, you have representatives from South Korea (Dunamu), China (DJI Innovations, GalaxySpace), and Australia (Canva). This proves that highly profitable, lean business models are succeeding on a global scale.

```python
WITH valuation_to_funding AS(
    SELECT 
        c.company,
        f.company_id,
        ROUND((f.valuation / 1000000.0), 2) AS valuation_in_millions,
        ROUND((f.funding / 1000000.0), 2) AS funding_in_millions,
        ROUND((f.valuation / 1000000.0) / (f.funding / 1000000.0), 2) AS Valuation_to_Funding_ratio
    FROM funding AS f
    INNER JOIN companies AS c USING(company_id)
    WHERE f.funding >= 0.01
),
percentile_table AS(
    SELECT
        PERCENTILE_CONT(0.25) WITHIN GROUP (ORDER BY Valuation_to_Funding_ratio) AS Q25_valuation,
        PERCENTILE_CONT(0.50) WITHIN GROUP (ORDER BY Valuation_to_Funding_ratio) AS Q50_valuation,
        PERCENTILE_CONT(0.75) WITHIN GROUP (ORDER BY Valuation_to_Funding_ratio) AS Q75_valuation
    FROM valuation_to_funding
)
SELECT 
    v.company_id,
    v.company,
    v.Valuation_to_Funding_ratio,
    CASE 
        WHEN v.Valuation_to_Funding_ratio < p.Q25_valuation THEN 'Capital Intensive'
        WHEN v.Valuation_to_Funding_ratio >= p.Q25_valuation AND v.Valuation_to_Funding_ratio < p.Q75_valuation THEN 'Standard'
        ELSE 'High Efficiency'
    END AS efficiency_tier 
FROM 
    valuation_to_funding AS v
-- CROSS JOIN attaches the single percentile row to every company row
CROSS JOIN 
    percentile_table AS p
ORDER BY 
    v.Valuation_to_Funding_ratio DESC;
```
